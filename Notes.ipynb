1. Relationsdatabaser (grunden)  
**Vad det är:**  
En databas där data lagras i tabeller (som Excel), med relationer mellan dem.

**Exempel:**  
- Tabell "Users"
- Tabell "Orders"
- Kopplas via user_id

**Varför du behöver det:**  
- All data i företag är strukturerad så här
- AI-modeller tränas ofta på sådan data
- Du måste förstå strukturen för att kunna analysera rätt  
  
2. SQL (det viktigaste verktyget)  
**Vad det är:**  
Ett språk för att prata med databaser.

**Exempel:**  
  
  
**Varför du behöver det:**  
- Hämta data för analys
- Filtrera, gruppera och räkna
- Absolut kärnkompetens inom data & AI

Utan SQL = du kommer inte åt datan.  
  
3. Databasoperationer (CRUD)  
**Vad det är:**  
Grundläggande saker du gör med data:  
- Create (lägga till)
- Read (läsa)
- Update (ändra)
- Delete (ta bort)

**Varför du behöver det:**   
- Bygga system som lagrar data
- Manipulera data innan analys
- Vanligt i ALL backend och datahantering  
  
4. Informationsstrukturering  
**Vad det är:**  
Hur man organiserar data på ett logiskt sätt.

**Exempel:**  
- Dela upp data i flera tabeller
- Undvika duplicering

**Varför du behöver det:**  
- Gör databasen snabbare och renare
- Viktigt för skalbara system
- AI kräver bra datakvalitet

Dålig struktur = dåliga analyser.  
  
5. Views, index och constraints  
**Vad det är:**
- Views: Sparade queries (som virtuella tabeller)
- Index: Gör databasen snabbare
- Constraints: Regler (t.ex. "ID måste vara unikt")

**Varför du behöver det:**
- Snabbare queries
- Säkrare data
- Viktigt i stora system  
  
6. Säkerhet i databaser  
**Vad det är:**
Hur man skyddar data.
  
**Exempel:**
- Behörigheter
- SQL injections

**Varför du behöver det:**
- Skydda känslig data
- Undvika attacker
- Lagkrav (GDPR m.m.)

# Grammatisk Ordning för SQL-kommandon

1. SELECT kolumner  - (väljer vilka kolumner (Ex "i tabellen kurser") man ska ta ut, som sedan visas i en tabell, med de rader som träffar de villkor som sätts upp (följande kolumner))  
2. FROM tabell   - Plocka från den tabellen. (vilket tabellnamn ex: "kurser",). Denna känns osm att den ligger över ovan steg i hierarkin, men det ska vara så. 
3. JOIN annan_tabell ON villkor - (Ofta vill man använda flera tabeller samtidigt, tex, namn på studerande som finns i annan tabell.)
4. WHERE radvillkor - (Vilka studenter börjar på "A") "Rader som innehåller detta". 
5. GROUP BY grupperingskolumner - Gruperar raderna utifrån villkor, ex poäng. Måste vara aggregerad, i select välja någon form av exempelvis "count". 
6. HAVING gruppvillkor - Som where fast för det man har grupperats. (ytterligare filtrering)
7. ORDER BY sortering - vad vi ska sortera efter 
8. LIMIT antal      - Hur många man är intresserad av att se. Ex topp 3.

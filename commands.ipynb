# SQL commands 

## Exaple of commands

- SELECT column1, column2, *
- INTO new_table
- FROM table_name
- WHERE condition
- UPDATE table_name
- SET column = value
- INSERT INTO table_name (columns)
- VALUES (values)
- DELETE FROM table_name
- GROUP BY column, COUNT(*)
- HAVING condition
- JOIN table2 ON condition
- CASE WHEN condition THEN value ELSE value END LTRIM(column)
- RTRIM(column)
- GO
- DROP TABLE table_name

## Used to

- Retrieving data → SELECT
- Creating tables → SELECT INTO
- Filtering → WHERE
- Updating → UPDATE
- Deleting → DELETE
- Inserting → INSERT
- Aggregating → GROUP BY, COUNT, HAVING
- Joining tables → JOIN
- Logic → CASE
- Structuring scripts → GO

## Commands explained

### SELECT – retrieve data (read data from a table) 
**Used to:**
- explore data
- verify results
- build other queries  
  
**"*"** in SELECT?
"all columns" 

```python
-- Syntax:
SELECT column1, column2
FROM table_name;

-- Example:
SELECT Spacecraft, Operator
FROM MoonMissions;
```

### SELECT INTO – create a new table from data
**Used to:**
- Create a new table
- Fill table with the result of a query  

```python
-- Syntax: 
SELECT columns
INTO new_table
FROM existing_table
WHERE condition;

-- Example:
SELECT *
INTO SuccessfulMissions
FROM MoonMissions
WHERE Outcome = 'Successful';
```

### WHERE – filter rows
**used to:**
- Filter which rows to include

```python
-- Syntax:
SELECT *
FROM table
WHERE condition; 

-- Example:
SELECT *
FROM MoonMissions 
WHERE Outcome = 'Successful'
-- Only returns outcome=successful missions
```

### UPDATE – modify data
**Used to:**  
- Update values in a table  
- clean data
- fix errors

```python
-- Syntax
UPDATE table_name
SET column = new_value;

-- Example: 
UPDATE SuccessfulMissions
SET Operator = LTRIM(RTRIM(Operator));
```

### INSERT – add new rows
**Used to:** 
- Add new data to a table

```python
-- Syntax: 
INSERT INTO table_name (column1, column2)
VALUES (value1, value2);

-- Example:
INSERT INTO NewUsers (Name, Gender)
VALUES ('Anna Andersson', 'Female');
```

### DELETE – remove rows
**Used to:**
- Delete rows from a table

```python
-- Syntax:
DELETE FROM table_name
WHERE condition;

-- Example:
DELETE FROM NewUsers
WHERE Gender = 'Female' AND BirthYear < 1970;
```

### GROUP BY – group data
**Used to:**
- Group rows for aggregation

```python
-- Syntax:
SELECT column, COUNT(*)
FROM table
GROUP BY column;

-- Example:
SELECT Operator, [Mission type], COUNT(*) AS MissionCount
FROM SuccessfulMissions
GROUP BY Operator, [Mission type];
```

### COUNT() – count rows
**Used to:**
- Count number of rows

```python
-- Syntax:
COUNT(*)

-- Example:
COUNT(*) AS MissionCount
```

### HAVING – filter grouped data
**Used to:**
- Filter results after grouping

```python
-- Syntax:
HAVING COUNT(*) > 1

-- Example:
HAVING COUNT(*) > 1
-- Only shows groups with more than one row
```

### JOIN – combine tables
**Used to:**
- Combine data from multiple tables

```python
-- Syntax:
SELECT *
FROM table1
JOIN table2 ON table1.id = table2.id;

-- Example:
SELECT p.ProductName, s.CompanyName
FROM Products p
JOIN Suppliers s ON p.SupplierID = s.SupplierID;
```

### CASE WHEN – conditional logic
**Used to:**
- Acts like an if/else statement in SQL

```python
-- Syntax:
CASE 
    WHEN condition THEN value
    ELSE value
END

-- Example:
CASE 
    WHEN PersonNumber % 2 = 0 THEN 'Female'
    ELSE 'Male'
END
```

### LTRIM() / RTRIM() – clean text
**Used to:**
- Remove whitespace
- clean messy data

```python
-- Syntax:
LTRIM(RTRIM(column))
```

### GO – separate execution blocks
**Used to:**
- Split your script into separate execution steps

```python
-- Example:
SELECT ...

GO

UPDATE ...
```

### DROP TABLE – delete a table
**Used to:**
- Remove a table completely

```python
-- Example:
DROP TABLE table_name;
```
